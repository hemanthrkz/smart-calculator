```python
import pandas as pd

import numpy as np



from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.metrics.pairwise import cosine_similarity

movies = {

    "title": [

        "Avengers",

        "Iron Man",

        "Captain America",

        "Thor",

        "The Dark Knight",

        "Batman Begins",

        "Spider-Man",

        "Interstellar",

        "Inception",

        "The Matrix",

        "Titanic",

        "The Notebook",

        "Jurassic Park",

        "Transformers",

        "Fast and Furious"

    ],



    "genres": [

        "action superhero adventure",

        "action superhero science fiction",

        "action superhero adventure",

        "action superhero fantasy",

        "action superhero crime",

        "action superhero crime",

        "action superhero adventure",

        "science fiction adventure drama",

        "science fiction action thriller",

        "science fiction action",

        "romance drama",

        "romance drama",

        "science fiction adventure",

        "action science fiction",

        "action racing adventure"

    ]

}



movies = pd.DataFrame(movies)



movies

tfidf = TfidfVectorizer(

    stop_words="english"

)



tfidf_matrix = tfidf.fit_transform(

    movies["genres"]

)



print("TF-IDF Matrix Shape:", tfidf_matrix.shape)

similarity = cosine_similarity(tfidf_matrix)



print("Similarity Matrix Shape:", similarity.shape)

def recommend_movies(movie_name, number_of_movies=5):



    movie_name = movie_name.lower()



    matches = movies[

        movies["title"].str.lower() == movie_name

    ]



    if matches.empty:

        print("Movie not found!")

        return



    movie_index = matches.index[0]



    similarity_scores = list(

        enumerate(similarity[movie_index])

    )



    similarity_scores = sorted(

        similarity_scores,

        key=lambda x: x[1],

        reverse=True

    )



    recommended_movies = []



    for index, score in similarity_scores[1:number_of_movies+1]:



        recommended_movies.append({

            "Movie": movies.iloc[index]["title"],

            "Similarity Score": round(score, 3)

        })



    result = pd.DataFrame(recommended_movies)



    return result

recommend_movies("Avengers")

movie = input("Enter a movie name: ")



result = recommend_movies(movie)



if result is not None:

    print("\n🎬 Recommended Movies:\n")

    display(result)

```
